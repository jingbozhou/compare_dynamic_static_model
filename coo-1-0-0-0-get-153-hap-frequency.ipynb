```python
"""
根据151水稻功能基因，提取对应6000份基因的频率
"""
import os
import numpy as np
import pandas as pd

HAPPER_PATH = "/share/home/zhoujb/project/riceBigExpBreeding/processData/2-allGeneHapPer/"
```

```python
info_data_path = "/share/home/zhoujb/project/riceDataComb/rawData/gene_coexpression_info.txt"
info_data = pd.read_table(info_data_path)
info_data["MSU"] = info_data["MSU_id"].str.split(".").str[0]

info_data = info_data.dropna(subset="MSU")

msu_set = set(info_data["MSU"].to_list())

info_data

# 获取目录下文件对应的基因ID列表
inte_list = [x.split(".")[0] for x in os.listdir(HAPPER_PATH) if x.split(".")[0] in msu_set]

# 一次性读取所有文件并放入字典
df_dict = {}
for item in inte_list:
    file_path = os.path.join(HAPPER_PATH, f"{item}.txt")
    tmp = pd.read_table(file_path, index_col="sample")[["percentage"]]
    tmp.columns = [item]  # 重命名列
    df_dict[item] = tmp

# 一次性 concat
df_per = pd.concat(df_dict.values(), axis=1)

print(df_per.shape)
```
