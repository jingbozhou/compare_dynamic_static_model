```python
"""
获取对应的表型信息
"""

import os
import numpy as np
import pandas as pd
```

```python
header = pd.read_excel("../rawData/Phenotype.xlsx", header=None, nrows=3)
header = header.fillna(method="ffill", axis=1)

data = pd.read_excel("../rawData/Phenotype.xlsx", header=None, skiprows=3)

cols = pd.MultiIndex.from_arrays(header.values)
data.columns = cols

df_long = data.melt(id_vars=[('ID', None, None)])
df_long.columns = ["ID", "trait", "location", "year", "value"]

df_long.to_csv("../processData/9-0-runOnline/phen_long.txt", sep="\t", index=0)

df_long["trait"] = (
    df_long["trait"]
    .str.replace(r"\s*\(.*?\)", "", regex=True)  # 去括号
    .str.strip()
    .str.replace(r"\s+", "_", regex=True)        # 多空格变_
)

df_matrix = df_long.pivot_table(
    index=["ID", "location", "year"],
    columns="trait",
    values="value"
)

df_matrix = df_matrix.reset_index()
df_matrix.to_csv("../processData/9-0-runOnline/phen_matrix.txt", sep="\t", index=0)
```
