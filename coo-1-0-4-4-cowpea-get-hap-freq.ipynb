```python
"""
根据151水稻功能基因，提取对应300份基因的频率
"""
import os
import numpy as np
import pandas as pd

HAPPER_PATH = "/share/home/zhoujb/project/riceDataComb/processData/dataFromCowpea/2-allGeneHapPer/"
```

```python
inte_list = [x.split(".")[0] for x in os.listdir(HAPPER_PATH)]

# 一次性读取所有文件并放入字典
df_dict = {}
for item in inte_list:
    file_path = os.path.join(HAPPER_PATH, f"{item}.txt")
    tmp = pd.read_table(file_path, index_col="sample")[["percentage"]]
    tmp.columns = [item]  # 重命名列
    df_dict[item] = tmp

# 一次性 concat
df_per = pd.concat(df_dict.values(), axis=1)

df_per.to_csv("../processData/dataFromCowpea/cowpea_hap_per.txt", sep="\t")

print(df_per.shape)
```
